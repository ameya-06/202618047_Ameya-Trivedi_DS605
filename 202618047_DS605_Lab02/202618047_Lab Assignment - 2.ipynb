Task 2 - Vectorized Arithmetic and Linear Algebra 

1. Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ 
or np.matmul(). 

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

#Matrix Addition
addition = A + B
print("Matrix Addition:")
print(addition)

#Element-wise Multiplication
element_multiplication = A * B
print("Element-wise Multiplication:")
print(element_multiplication)

#Matrix Multiplication using @
matrix_multiplication = A @ B
print("Matrix Multiplication:")
print(matrix_multiplication)
```

2. For a square matrix, find its transpose and determinant

```python
transpose_A = A.T
print("Transpose of A:")
print(transpose_A)

det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

```

3. If invertible, find the inverse and verify A @ A^-1 using np.allclose(). 

```python
inverse_A = np.linalg.inv(A)
print("Inverse of A:")
print(inverse_A)

#Verify A × A⁻¹ using np.allclose()
identity_check = A @ inverse_A
print("A @ A^-1:")
print(identity_check)

print("Is it approximately an identity matrix?",
      np.allclose(identity_check, np.eye(2)))
```

3. Try to fill NaN values by imputing using: mean, median, mode, random value 

```python
# Mean imputation
df_mean = pd.read_csv("train.csv")
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

print("Mean imputation - missing Age:",
      df_mean["Age"].isnull().sum())


# Median imputation
df_median = pd.read_csv("train.csv")
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

print("Median imputation - missing Age:",
      df_median["Age"].isnull().sum())


# Mode imputation
df_mode = pd.read_csv("train.csv")

age_mode = df_mode["Age"].mode()[0]

df_mode["Age"] = df_mode["Age"].fillna(age_mode)

print("Age Mode:", round(age_mode, 2))
print("Mode imputation - missing Age:",
      df_mode["Age"].isnull().sum())


# Random value imputation
df_random = pd.read_csv("train.csv")

np.random.seed(42)

missing_age = df_random["Age"].isnull()

df_random.loc[missing_age, "Age"] = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_age.sum()
)

print("Random imputation - missing Age:",
      df_random["Age"].isnull().sum())
```

4. For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers. 

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))

#IQR
IQR = Q3 - Q1

print("IQR:", round(IQR, 2))\

#1.5 × IQR bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))

# Finding outliers 
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Number of Fare outliers:", len(fare_outliers))
```

2. Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest 
groups. 

```python
pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("Survival Rate Pivot Table:")
print((pivot_table * 100).round(2))

pivot_percentage = (pivot_table * 100).round(2)

highest_group = pivot_percentage.stack().idxmax()
highest_value = pivot_percentage.stack().max()

lowest_group = pivot_percentage.stack().idxmin()
lowest_value = pivot_percentage.stack().min()

print("\nHighest survival group:")
print(highest_group, "=", round(highest_value, 2), "%")

print("\nLowest survival group:")
print(lowest_group, "=", round(lowest_value, 2), "%")
```

Task 9 - Visualizations and Observations 

1. Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative 
relationships. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")

# Select relevant numerical columns
numeric_columns = [
    "Survived", "Pclass", "Age",
    "SibSp", "Parch", "Fare"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Titanic Numerical Variables")
plt.tight_layout()
plt.show()

# Find strongest positive and negative correlations
corr_pairs = correlation.unstack()
corr_pairs = corr_pairs[corr_pairs != 1]

print("Strongest positive correlation:")
print(corr_pairs.idxmax(), "=", round(corr_pairs.max(), 2))

print("\nStrongest negative correlation:")
print(corr_pairs.idxmin(), "=", round(corr_pairs.min(), 2))
```
